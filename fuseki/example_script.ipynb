```python
from SPARQLWrapper import SPARQLWrapper, JSON
```

```python
FUSEKI_URL = "http://localhost:3030/pokemon/sparql"
sparql = SPARQLWrapper(FUSEKI_URL)
```

#### Just an example query to print pokemons starting with p, have set appropriate limit to include pikachu :P

```python
query_string = """
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX poke: <https://pokemonkg.org/ontology#>

SELECT DISTINCT ?name
FROM <urn:x-arq:UnionGraph>
WHERE {
    ?pokemon a poke:Species .
    ?pokemon rdfs:label ?name .
    FILTER regex(?name, "^P", "i")
    FILTER (lang(?name) = "en")
}
ORDER BY ?name
LIMIT 30
"""
```

```python
sparql.setQuery(query_string)
sparql.setReturnFormat(JSON)

try:
    results = sparql.query().convert()
    bindings = results["results"]["bindings"]

    if not bindings:
        print("No results, check if UnionGraph is active")
    else:
        for result in bindings:
            name = result["name"]["value"]
            print(name)
except Exception as e:
    print(f"An error occured: {e}")
```
